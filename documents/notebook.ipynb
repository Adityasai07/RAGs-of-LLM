```python
### Read all the pdf's inside the directory
def process_all_pdfs(pdf_directory):
    """Process all PDF files in a directory"""
    all_documents = []
    pdf_dir = Path(pdf_directory)
    
    # Find all PDF files recursively
    pdf_files = list(pdf_dir.glob("**/*.pdf"))
    
    print(f"Found {len(pdf_files)} PDF files to process")
    
    for pdf_file in pdf_files:
        print(f"\nProcessing: {pdf_file.name}")
        try:
            loader = PyPDFLoader(str(pdf_file))
            documents = loader.load()
            
            # Add source information to metadata
            for doc in documents:
                doc.metadata['source_file'] = pdf_file.name
                doc.metadata['file_type'] = 'pdf'
            
            all_documents.extend(documents)
            print(f"  ✓ Loaded {len(documents)} pages")
            
        except Exception as e:
            print(f"  ✗ Error: {e}")
    
    print(f"\nTotal documents loaded: {len(all_documents)}")
    return all_documents

# Process all PDFs in the data directory
all_pdf_documents = process_all_pdfs("../data")
```

```python
# chucks splits 
def split_documents(documents,chunk_size=1000,chunk_overlap=200):
    """Split documents into smaller chunks for better RAG performance"""
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
        separators=["\n\n", "\n", " ", ""]
    )
    split_docs = text_splitter.split_documents(documents)
    print(f"Split {len(documents)} documents into {len(split_docs)} chunks")
    
    # Show example of a chunk
    if split_docs:
        print(f"\nExample chunk:")
        print(f"Content: {split_docs[0].page_content[:200]}...")
        print(f"Metadata: {split_docs[0].metadata}")
    
    return split_docs
```

```python
import numpy as np 
from sentence_transformers import SentenceTransformer
import chromadb
from chromadb.config import Settings
import uuid 
from sklearn.metrics.pairwise import cosine_similarity 
from typing import Any , List , Tuple , Dict 
```

```python
class EmbeddingManager : 
    def __init__(self , model_name : str = "all-MiniLM-L6-v2"):
        self.model_name =  model_name
        self.model = None
        self._load_model()
    def _load_model(self) : 
        try : 
            print("Started to load model" , self.model_name)
            self.model = SentenceTransformer(self.model_name)
            print("Model Loaded successfully " , self.model_name)
        except Exception as e : 
            print(f"Error occured {self._load_model} with e : {e}")
            raise 
    def generate_embeddings(self , texts : list[str]) -> np.ndarray:
        if not self.model :
            raise ValueError("model not loaded yet !!!")
        print("Lenght of texts : " , len(texts))
        print("Started Embedding !!!")
        embeddings = self.model.encode(texts , show_progress_bar=True) 
        return  embeddings

    def get_dimensions(self):
        print(f"Dimensions : {self.model.get_embedding_dimension()}")
```

```python
# vector db 

class VectorStore() : 
    def __init__(self , collection_name : str = "pdf_docs" , db_directory : str = '../data/VectorStore'):
        self.collection_name = collection_name 
        self.db_directory = db_directory 
        self.client = None 
        self.collection = None 
        self._initilize_store()

    def _initilize_store(self) : 
        try : 
            os.makedirs(self.db_directory , exist_ok=True)
            self.client = chromadb.PersistentClient(path=self.db_directory)
            self.collection = self.client.get_or_create_collection(
                name = self.collection_name,
                metadata = {
                    "model_name": "all-MiniLM-L6-v2",
                    "description" : "for pdfs creating RAG"
                }
            )
            print("Collection : " , self.collection_name)
            print("Collections count : " , self.collection.count())
        except Exception as e :
            raise ValueError(f"Error Occured !!! {e}")

    def add_documents(self, documents : list[Any] , embeddings : np.ndarray):
        if len(documents) != len(embeddings) : 
            raise ValueError("Chunks and embeddings lenght doesnt match" , len(embeddings) , " - " , len(documents))
        print("Preparing for adding docs of lenght : " , len(embeddings))
        ids = []
        new_embeddings = []
        new_metadatas = [] 
        new_documents = []
        for i , (document , embedding) in enumerate(zip(documents , embeddings)) : 
            document_id = f"doc_{uuid.uuid4().hex[:8]}_{i}"
            ids.append(document_id)
            m = dict(document.metadata)
            m['doc_index'] = i 
            m['content_length'] = len(document.page_content)
            new_metadatas.append(m)
            new_documents.append(document.page_content)
            new_embeddings.append(embedding.tolist())

        try : 
            self.collection.add(
                ids = ids , 
                embeddings= new_embeddings , 
                metadatas= new_metadatas , 
                documents= new_documents
            )
            print("Successfully added to collections !!!")
            print("Colllections Count : " , self.collection.count())
        except Exception as e : 
            raise ValueError(f"Error Occured !!! {e}")
```

```python
class RAGretriver:
    def __init__(self , vector_store :  VectorStore, embedding_manager : EmbeddingManager):
        self.vector_store = vector_store 
        self.embedding_manager = embedding_manager
        
    def retrive(self , query , topk : int = 5 , threshold_score : float = 0.0) -> list[dict[str , Any]]:
        print("query : " , query , "\ttopk : " , topk,"\tThreshold score : " , threshold_score)
        embeddings_ofquery = self.embedding_manager.generate_embeddings([query])[0]
        try : 
            results_ofquery_from_vector_store = self.vector_store.collection.query(
                query_embeddings=[embeddings_ofquery.tolist()] , 
                n_results = topk
            )
            retrived_documents = []
            if results_ofquery_from_vector_store['documents'] and results_ofquery_from_vector_store['documents'][0] : 
                documents = results_ofquery_from_vector_store['documents'][0]
                metadatas = results_ofquery_from_vector_store['metadatas'][0]
                distances = results_ofquery_from_vector_store['distances'][0]
                ids = results_ofquery_from_vector_store['ids'][0]
                for i , (document_id , document , metadata , distance) in enumerate(zip(ids , documents , metadatas , distances)) : 
                    score = 1 - distance 
                    if score > threshold_score : 
                        retrived_documents.append(
                            {
                                'id' : document_id , 
                                'content' : document , 
                                'metadata' : metadata , 
                                'similarity_score' : score , 
                                'distance' : distance,
                                'rank' : len(retrived_documents) + 1 
                            }
                        )
                print(f"Retrived documents count : {len(retrived_documents)}")
            else : 
               print("No relevant docs found")

            return retrived_documents 
        except Exception as e : 
            print(f"Error occured : {e}")
            raise 
```

```python
def rag(query , rag_retriver, llm, top_k) : 
    results = rag_retriver.retrive(query=query , topk=top_k)
    context = "\n\n".join(
        [doc['content'] for doc in results]
    ) if results else ""
    if context :
        prompt = f"""Use the following context to answer the question.
Context:
{context}

Question:
{query}

Answer:"""
        response = llm.invoke(prompt)  
        return response.content if response.content else "No response from LLM."
    else : 
        return "No relevant documents found for the query."
```

```python
def rag_advance(query , rag_retriver , llm , top_k = 5 , min_score = 0.2 , return_context = False) :
    results = rag_retriver.retrive(query=query , topk = top_k , threshold_score= min_score)
    if not results :
        return {
            'answer' : "No relevant context found", 
            'sources' : [] , 
            'confidence' : 0.0 , 
            'context' : ''
        }

    context = "\n\n".join([doc['content'] for doc in results])
    sources = [{
        'source': doc['metadata'].get('source_file', doc['metadata'].get('source', 'unknown')),
        'page': doc['metadata'].get('page', 'unknown'),
        'score': doc['similarity_score'],
        'preview': doc['content'][:250] + '...'
    } for doc in results]
    confidence = max([doc['similarity_score'] for doc in results])

    prompt = f"""Use the following context to answer the question.\nContext:\n{context}\n\nQuestion: {query}\n\nAnswer:"""
    response = llm.invoke(prompt)
    output = {
        'answer' : response.content,
        'sources' : sources , 
        'confidence' : confidence 
    }
    if return_context :
        output['context'] = context 
    return output


```

```python
result = rag_advance("What is reinforcement learning?", rag_retriver, llm, top_k=3, min_score=0.1, return_context=True)
print("Answer:", result['answer'])
print("Sources:", result['sources'])
print("Confidence:", result['confidence'])
print("Context Preview:", result['context'][:300])
```

```python
from typing import List, Dict, Any
import time

class RAGpipeline:
    def __init__(self, rag_retriver, llm):
        self.rag_retriver = rag_retriver
        self.llm = llm
        self.history = []  # Store query history

    def query(self, query: str, top_k: int = 5, min_score: float = 0.2, stream: bool = False, summarize: bool = False) -> Dict[str, Any]:
        # Retrieve relevant documents
        results = self.rag_retriver.retrive(query , top_k , min_score )
        if not results:
            answer = "No relevant context found."
            sources = []
            context = ""
        else:
            context = "\n\n".join([doc['content'] for doc in results])
            sources = [{
                'source': doc['metadata'].get('source_file', doc['metadata'].get('source', 'unknown')),
                'page': doc['metadata'].get('page', 'unknown'),
                'score': doc['similarity_score'],
                'preview': doc['content'][:120] + '...'
            } for doc in results]
            # Streaming answer simulation
            prompt = f"""Use the following context to answer the question concisely.\nContext:\n{context}\n\nQuestion: {query}\n\nAnswer:"""
            if stream:
                print("Answer: ", end="", flush=True)
                streamed_tokens = []
                # llm.stream() yields AIMessageChunk objects as Groq generates tokens
                for chunk in self.llm.stream(prompt):
                    token = chunk.content
                    print(token, end="", flush=True)
                    streamed_tokens.append(token)
                print("\n")
                answer = "".join(streamed_tokens)
            else:
                response = self.llm.invoke(prompt)
                answer = response.content

        # Add citations to answer
        citations = [f"[{i+1}] {src['source']} (page {src['page']})" for i, src in enumerate(sources)]
        answer_with_citations = answer + "\n\nCitations:\n" + "\n".join(citations) if citations else answer

        # Optionally summarize answer
        summary = None
        if summarize and answer:
            summary_prompt = f"Summarize the following answer in 2 sentences:\n{answer}"
            summary_resp = self.llm.invoke(summary_prompt)
            summary = summary_resp.content

        # Store query history
        self.history.append({
            'query': query,
            'answer': answer,
            'sources': sources,
            'summary': summary
        })

        return {
            'query': query,
            'answer': answer_with_citations,
            'sources': sources,
            'summary': summary,
            'history': self.history
        }

```

```python
# Example usage:
adv_rag = RAGpipeline(rag_retriver , llm)
result = adv_rag.query("How was the Evaluation of Information Retrieval Systems", top_k=3, min_score=0.2, stream=True, summarize=True)
print("\nFinal Answer:", result['answer'])
print("Summary:", result['summary'])
print("History:", result['history'][-1])
```
